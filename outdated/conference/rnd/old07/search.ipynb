```python
import json
data_t2 = None
with open('t2_result_short.json', 'r') as file:
    data_t2 = json.load(file)
display(data_t2)
```

```python
import re
# https://stackoverflow.com/a/14537848
latex_formula_regex=r'\$.*\$'

def export_text(data: dict):
    text = ""
    for block in data["parsing_res_list"]:
        if block["block_label"] == "text":
            line = re.sub(latex_formula_regex, "<F>", block["block_content"])
            print(f"TEXT [{block["block_label"]}] ({line[0:10]}...)")
            text += line
        else:
            print(f"SKIP [{block["block_label"]}]")
    return text
text = export_text(data_t2)
print(text)
```

```python
def export_formulas(data: dict):
    formulas = []
    for block in data["formula_res_list"]:
        formulas.append(block["rec_formula"])
    return formulas
formulas = export_formulas(data_t2)
formulas
```

```python
import sympy
# Could not get it to work:  LaTeX parsing requires the antlr4 Python package, provided by pip (antlr4-python3-runtime) or conda (antlr-python-runtime), version 4.11
#from sympy.parsing.latex import parse_latex
from latex2sympy2_extended import latex2sympy
def normalize_formulas(formulas: list[str]):
    data = []
    for line in formulas:
        # Does not make sense
        if line.find('=') == -1 or line[0]=="=" or line[-1]=="=":
            print(f"[SKIP] Incorrect formula: {line}.")
            continue
        try:
            formula = latex2sympy(line)
            formula_simple = sympy.simplify(formula)
            if type(formula_simple) == bool:
                print(f"[OK] Parsed formula: {formula}. If was simplified: {formula_simple}. Was: {line}.")
                data.append(formula)
            else:
                print(f"[OK] Parsed formula: {formula_simple}. Not simplified: {formula}. Was: {line}.")
                data.append(formula_simple)
        except Exception as e:
            print(f"[FAIL] Failed to parse formula: {line}. Exception: {e}")
    return data
latex_formulas = normalize_formulas(formulas)
for i in latex_formulas:
    print(i)
```

```python
import latex2mathml.converter
def normalize_formulas_to_mathml(formulas: list[str]):
    data = []
    for line in formulas:
        # Does not make sense
        if line.find('=') == -1 or line[0]=="=" or line[-1]=="=":
            print(f"[SKIP] Incorrect formula: {line}.")
            continue
        try:
            formula = latex2mathml.converter.convert(line)
            print(f"[OK] Parsed formula: {formula}. Was: {line}.")
            data.append(formula)
        except Exception as e:
            print(f"[FAIL] Failed to parse formula: {line}. Exception: {e}")
    return data
mathml_formulas = normalize_formulas_to_mathml(formulas)
for i in mathml_formulas:
    print(i)
```
