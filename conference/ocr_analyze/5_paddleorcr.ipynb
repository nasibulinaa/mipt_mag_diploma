```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(
    use_doc_orientation_classify=False, 
    use_doc_unwarping=False, 
    use_textline_orientation=False) # text detection + text recognition
# ocr = PaddleOCR(use_doc_orientation_classify=True, use_doc_unwarping=True) # text image preprocessing + text detection + textline orientation classification + text recognition
# ocr = PaddleOCR(use_doc_orientation_classify=False, use_doc_unwarping=False) # text detection + textline orientation classification + text recognition
# ocr = PaddleOCR(
#     text_detection_model_name="PP-OCRv5_mobile_det",
#     text_recognition_model_name="PP-OCRv5_mobile_rec",
#     use_doc_orientation_classify=False,
#     use_doc_unwarping=False,
#     use_textline_orientation=False) # Switch to PP-OCRv5_mobile models
result = ocr.predict("../../data/for_rag/t3.pdf")
for res in result:
    res.print()
    res.save_to_img("paddleocr_output")
    res.save_to_json("paddleocr_output")
```

```python
from paddleocr import TextRecognition

model = TextRecognition()
output = model.predict(input="../../data/for_rag/t3.pdf")
for res in output:
    res.print()
    res.save_to_img(save_path="paddleocr_output1")
    res.save_to_json(save_path="paddleocr_output1/res.json")
```

```python
from paddleocr import TextDetection

model = TextDetection()
output = model.predict("../../data/for_rag/t3.pdf")
for res in output:
    res.print()
    res.save_to_img(save_path="./paddleocr_output2/")
    res.save_to_json(save_path="./paddleocr_output2/res.json")
```

```python
from paddleocr import PPStructureV3

pipeline = PPStructureV3(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
)
output = pipeline.predict(
    input="../../data/for_rag/t3.pdf")
for res in output:
    res.print()
    res.save_to_json(save_path="paddleocr_output3")
    res.save_to_markdown(save_path="paddleocr_output3")
```

```python
from paddleocr import PPStructureV3

pipeline = PPStructureV3(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    lang="ru"
)
output = pipeline.predict(
    input="../../data/for_rag/t3.pdf")
for res in output:
    res.print()
    res.save_to_json(save_path="paddleocr_output3_ru")
    res.save_to_markdown(save_path="paddleocr_output3_ru")
    res.save_to_img(save_path="./paddleocr_output3_ru/")
```

```python
from IPython.display import display, Math, Image
for res in output:
    print(res["input_path"])
    for form in res.str['res']['formula_res_list']:
        print(f"Formula: {form["rec_formula"]}")
        display(Math(form["rec_formula"]))
```
